```python
import sqlite3

# Đọc file truy vấn và tách thành từng cặp (mục đích, truy vấn)
def read_queries(filename):
    queries = []
    with open(filename, encoding='utf-8') as f:
        lines = f.readlines()
    purpose = ""
    query = ""
    for line in lines:
        if line.strip().startswith('--'):
            if query.strip():
                queries.append((purpose.strip(), query.strip()))
                query = ""
            purpose = line.strip().lstrip('--').strip()
        elif line.strip():
            query += line
    if query.strip():
        queries.append((purpose.strip(), query.strip()))
    return queries

# Kết nối tới database
conn = sqlite3.connect('Data.db')  # Thay bằng đường dẫn DB của bạn
cur = conn.cursor()

queries = read_queries('querry.sql')

for idx, (purpose, sql) in enumerate(queries, 1):
    print(f"{idx}. Mục đích: {purpose}")
    print(f"Truy vấn: {sql}")
    try:
        cur.execute(sql)
        rows = cur.fetchall()
        if rows:
            for row in rows:
                print(row)
        else:
            print("Không có kết quả.")
    except Exception as e:
        print(f"Lỗi: {e}")
    print('-' * 50)

cur.close()
conn.close()
```
